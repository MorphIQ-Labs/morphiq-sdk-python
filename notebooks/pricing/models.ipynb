# MorphIQ SDK for Python — Available models

Inspect model capabilities, aliases, constraints, engine versions and Greek units.
Run setup once, then click **Load models — 1 request, 0 CU**. No key is needed.
Setup makes no requests. This catalog describes compiled capabilities; use the
cost quotation workspace for account-specific rates and admission.

[Back to start](../start-here.ipynb)


```python
import ipywidgets as widgets
from IPython.display import display

from morphiq_sdk import MorphIQ, MorphIQError

load_button = widgets.Button(
    description="Load models — 1 request, 0 CU", layout=widgets.Layout(width="300px")
)
output = widgets.Output()
last_catalog = None


def load_models(_):
    global last_catalog
    last_catalog = None
    load_button.disabled = True
    with output:
        output.clear_output(wait=True)
        try:
            with MorphIQ() as client:
                last_catalog = client.pricing_models()
            for model in last_catalog.models:
                print(model.name, "—", model.id, "—", model.model_version)
                print("Single:", ", ".join(model.operations))
                print("Batch:", ", ".join(model.batch_operations) or "none")
                print("Aliases:", model.aliases)
                print("Guidance:", model.guidance)
                for constraint in model.constraints:
                    print("Constraint:", constraint.model_dump(exclude_unset=True))
                print()
            print("Greek units:", last_catalog.greek_units)
            print("Engines:", [engine.model_dump() for engine in last_catalog.engines])
        except MorphIQError as error:
            print(error)
        finally:
            load_button.disabled = False


load_button.on_click(load_models)
display(load_button, output)
```
